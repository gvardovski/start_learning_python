```python
import pandas as pd
import pandas_ta as ta
import vectorbt as vbt

df = pd.read_csv("MSFT_1hour.csv", index_col="Time")
df.index = pd.to_datetime(df.index)
pf_rsi = {}
for i in ['7','14','21']:
    df[f'RSI_{i}'] = ta.rsi(df['Close'], length=int(i))

    entries = df[f'RSI_{i}'] < 30
    exits = df[f'RSI_{i}'] > 70

    pf = vbt.Portfolio.from_signals(
    entries = entries,
    exits = exits,
    open = df["Open"],
    close = df["Close"],
    high = df["High"],
    low = df["Low"],
    size = 1,
    size_type ="amount",
    fees=0.0035,
    slippage=0.2,
    init_cash=1000,
    freq ="1hour"
    )
    pf_rsi.update({f'RSI_{i}':pf})
```

```python
pf_rsi['RSI_14'].stats()
```

```python
pf_rsi['RSI_14'].plot()
```

```python
import pandas as pd
import pandas_ta as ta
import vectorbt as vbt

df = pd.read_csv("MSFT_1hour.csv", index_col="Time")
df.index = pd.to_datetime(df.index)
for i in ['50','200']:
    df[f'SMA_{i}'] = ta.sma(df['Close'], length=int(i))
entries = (df['SMA_50'] > df['SMA_200']) & (df['SMA_50'].shift(1) <= df['SMA_200'].shift(1))
exits = (df['SMA_50'] < df['SMA_200']) & (df['SMA_50'].shift(1) >= df['SMA_200'].shift(1))
pf = vbt.Portfolio.from_signals(
    entries = entries,
    exits = exits,
    open = df["Open"],
    close = df["Close"],
    high = df["High"],
    low = df["Low"],
    size = df['Volume'],
    size_type ="amount",
    fees=0.0035,
    slippage=0.2,
    init_cash=1000,
    freq ="1hour"
    )
```

```python
pf.stats()
```

```python
pf.plot()
```
